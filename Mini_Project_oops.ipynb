**📚 Project 1: Student Management System**

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
        self.marks = []

    def store_marks(self, math, science, english):
        """Stores student marks in a list."""
        self.marks = [math, science, english]

    def calculate_total(self):
        """Returns the sum of all marks."""
        return sum(self.marks)

    def calculate_average(self):
        """Returns the average score."""
        return sum(self.marks) / len(self.marks) if self.marks else 0

    def calculate_grade(self):
        """Determines the grade based on the average score."""
        avg = self.calculate_average()
        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        else:
            return "F"

    def display_report(self):
        """Prints a comprehensive performance report."""
        print("\n" + "="*30)
        print(f"      STUDENT REPORT CARD      ")
        print("="*30)
        print(f"Name        : {self.name}")
        print(f"Roll Number : {self.roll_number}")
        if self.marks:
            print("-"*30)
            print(f"Marks -> Maths   : {self.marks[0]}")
            print(f"Marks -> Science : {self.marks[1]}")
            print(f"Marks -> English : {self.marks[2]}")
            print("-"*30)
            print(f"Total Marks  : {self.calculate_total()}")
            print("-"*30)
            print(f"Average      : {self.calculate_average():.2f}")
            print("-"*30)
            print(f"Grade        : {self.calculate_grade()}")
        else:
            print("No marks recorded yet.")
        print("="*30)
        print()
        print("X"*30)


# Demonstration
student1 = Student("Alice Smith", "S101")
student1.store_marks(92, 88, 95)
student1.display_report()

student2= Student("Luci Ben", "S223")
student2.store_marks(99, 88, 77)
student2.display_report()
```

**🏦 Project 2: Bank Account Management**

```python
class BankAccount:
    def __init__(self, account_holder, opening_balance=0.0):
        self.account_holder = account_holder
        self.balance = float(opening_balance)

    def deposit(self, amount):
        """Adds funds to the account."""
        if amount > 0:
            self.balance += amount
            print(f"Successfully deposited ${amount:.2f} into {self.account_holder}'s account.")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        """Deducts funds if adequate balance is available."""
        if amount > self.balance:
            print(f"❌ Withdrawal Failed: Insufficient funds. Current balance: ${self.balance:.2f}")
        elif amount <= 0:
            print("Invalid withdrawal amount.")
        else:
            self.balance -= amount
            print(f"Successfully withdrew ${amount:.2f} from {self.account_holder}'s account.")

    def check_balance(self):
        """Displays current account holdings."""
        print(f"Account Holder: {self.account_holder} | Current Balance: ${self.balance:.2f}")


# Demonstration (Maintaining independent balances)
account_a = BankAccount("John Doe", 500.0)
account_b = BankAccount("Jane Smith", 1000.0)

account_a.deposit(200)
account_a.withdraw(150)
account_a.check_balance()

account_b.withdraw(1200)  # Triggers insufficient funds
account_b.check_balance()  # Balance remains completely unaffected by account_a's activity
```

**💼 Project 3: Employee Payroll System**

```python
class Employee:
    def __init__(self, employee_id, name, basic_salary):
        self.employee_id = employee_id
        self.name = name
        self.basic_salary = float(basic_salary)

    def calculate_hra(self):
        """HRA = 20% of basic salary"""
        return 0.20 * self.basic_salary

    def calculate_da(self):
        """DA = 10% of basic salary"""
        return 0.10 * self.basic_salary

    def calculate_pf(self):
        """PF Deduction = 12% of basic salary"""
        return 0.12 * self.basic_salary

    def calculate_net_salary(self):
        """Net Salary = Basic Salary + HRA + DA - PF"""
        hra = self.calculate_hra()
        da = self.calculate_da()
        pf = self.calculate_pf()
        return self.basic_salary + hra + da - pf

    def display_salary_slip(self):
        """Prints a detailed break-down of the monthly payroll item."""
        print("\n" + "="*35)
        print(f"         PAYROLL SLIP          ")
        print("="*35)
        print(f"ID          : {self.employee_id}")
        print(f"Name        : {self.name}")
        print("-"*35)
        print(f"Base Salary : ${self.basic_salary:.2f}")
        print(f"HRA (+)     : ${self.calculate_hra():.2f}")
        print(f"DA (+)      : ${self.calculate_da():.2f}")
        print(f"PF (-)      : ${self.calculate_pf():.2f}")
        print("_"*35)
        print(f"NET SALARY:  ${self.calculate_net_salary():.2f}")
        print("_"*35)


# Demonstration
emp1 = Employee("EMP007", "Robert Chen", 5000.0)
emp1.display_salary_slip()

emp2 = Employee("EMP034", "Alice Ben", 8000.0)
emp2.display_salary_slip()

```
