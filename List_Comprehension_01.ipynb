```python
# Total Sales    , Name
sales = [
    ["Aman", 12000, 15000, 18000, 22000],
    ["Priya", 20000, 17000, 14000, 20000],
    ["Rohit", 16000, 22000, 19000, 18000]
]
highest_sale=0
name=""

for person in sales :
    total = sum(person[1:])
    name = person[0]

    print("Name :",name ) 
    print("Total Sale :",total)
    print("-----------------------------")
if total>highest_sale :
    print("Top Salesperson :",name)
    print("Highest sale :",total)
    
```
