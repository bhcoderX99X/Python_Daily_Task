```python
# Question 1. ATM Amount Withdraw
balance=20000
pin=input("Enter Your PIN: ")
if pin=="12345":
    amount=int(input("Enter Your Amount do you want to Withdraw: "))
    if(amount%100==0):
        if(balance>amount):
            balance -= amount          #balance = balance - amount
            print("Withdraw Amount is: ", amount)
            print("Remainig Balance is: ", balance)
        else:
            print("Insuficient Balance ")
    else:
        print("Invalid Amount ")
else:
    print("Invalid PIN ")
    
```

```python
# Question 2. Student Result System 
per = int(input("Enter Student Percentage(%) : "))
if(per >= 40):
    if(per >= 75):
        print("Distinction")
    elif(per >= 60):
        print("First Division")
    elif(per >= 50):
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

```python
# Question 3. Bank Loan Eligibility
age = int(input("Enter Your Age :"))
if(age >= 21):
    salary = float(input("Enter Your Monthly Salary :"))
    if(salary >= 25000):
        credit_score = int(input("Enter Your Credit Score :"))
        if(credit_score >= 750):
            print("Status : Loan Approved")
        elif(credit_score >= 650):
            print("Status : Loan Approved with higher interest")
        else:
            print("Status : Loan Rejected (Credit Score must be greater than or equal to 650)")
    else:
        print("Monthly salry is low (Salary must be greater than or equal 25000 )")
else:
    print("Invalid Age (Age must be 21 years or older)")
```

```python
# Question 5. Login = Role Based Access
user_name=input("Enter Your User Name: ")
if(user_name=="Admin" or user_name=="admin"):
    password=input("Enter Your Password:")
    if(password=="12345"):
        role=input("Enter User Role: ")
        if(role=="Admin" or "admin"):
            print("Full Access")
        elif(role=="Trainer"):
            print("Trainer Dashboard")
        elif(role=="Student"):
            print("Student Dashboard")
        else:
            print("Invalid User Name")
    else:
        print("Wrong Password")
else:
    print("Invalid User Name")
            
```

```python
# Question 6. Movie Ticket Booking
seat_avl=int(input("Available seat number :"))
if(seat_avl>0):
    age=int(input("Enter Age :"))
    if(age<5):
        print("Ticket is free of cost for age below 5")
    elif(age>=5 and age<=12):
        print("50% Discount on Ticket Price for age 5 to 12")
    elif(age>60):
        print("30% Discount on Ticket Price for age above 60")
    else:
        print("Full Price (No Discount)")
else:
    print("House Full (no seats are available)")
```

```python
# Student Grading System
python=int(input("Enter Your Python Marks :"))
if(python>=40):
    java=int(input("Enter Your Java Marks :"))
    if(java>=40):
        dbms=int(input("Enter Your DBMS Marks :"))
        if(dbms>=40):
            total_marks = python + java + dbms
            percentage = total_marks/3
            if(percentage>=90):
                print("Student Percentage is :",percentage)
                print("Grade A+")
            elif(percentage>=80):
                print("Student Percentage is :",percentage)
                print("Grade A")
            elif(percentage>=70):
                print("Student Percentage is :",percentage)
                print("Grade B")
            elif(percentage>=60):
                print("Student Percentage is :",percentage)
                print("Grade C")
            elif(percentage>=50):
                print("Student Percentage is :",percentage)
                print("Grade D")
            else:
                print("Grade Fail")
        else:
            print("Failed in DBMS")
    else:
        print("Failed in Java")
else:
    print("Failed in Python")
        
    
```
