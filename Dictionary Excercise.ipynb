```python
# Question->1. Check Student result
students = {
    "Rahul" : 75,   #item
    "Amit" : 32,
    "Priya" : 88,
    "Neha" : 39,
    "Rohit" : 55
}
for name, marks in students.items():
    if(marks>40):
        print(name, "Pass")
    else:
        print(name, "Fail")
```

```python
# Question->2. Student Performance Using Nested If
student={
    "Rahul" : 75,   #item
    "Amit" : 32,
    "Priya" : 88,
    "Neha" : 39,
    "Rohit" : 55
}
for name, marks in student.items():
    if(marks>40):
        if(marks>=80):
            print(name, "Excellent")
        elif(marks>=60):
            print(name, "good")
        elif(marks>=40):
            print(name, "avg")
    else:
        print(name, "fail")
    
               
    
```

```python
#Quetion->3. Product Stock Management
products={
    "Laptop" : 10,
    "Mouse" : 3,
    "Keyboard" : 0,
    "Monitor" : 7,
    "Printer" : 2
}
for pro_name,stock in products.items():
    if(stock>0):
        if(stock<=5):
            print(pro_name,"low stock")
        elif(stock>5):
            print(pro_name,"available")
    else:
        print(pro_name, "Out of stock")
        
```

```python
#Question->4. Display the Employee Category
employees = {
    "Rahul": {"salary" : 60000, "department" : "IT"},
    "Amit": {"salary" : 40000, "department" : "HR"},
    "Priya": {"salary" : 70000, "department" : "IT"},
    "Neha": {"salary" : 45000, "department" : "Finance"}
}
for name, info in employees.items():
    salary=info["salary"]
    dept=info["department"]
    if(salary>=50000):
        if("department"=="IT"):
            print(name, " is Senior IT Employees")
        elif("department"=="HR"):
            print(name, " is Senior HR Employees")
        else:
            print(name, " is Senior Employee")
    else:
        print(name, " is a Junior Employee")
        
        
    
```

```python
#Question->5. check Even/Odd values
data = {
    "a" : 10,
    "b" : 15,
    "c" : 25,
    "d" : 30,
    "e" : 40
}
for key, values in data.items():
    if(values%2==0):
        if(values>=20):
            print(key,"Large Even")
        elif(values<=20):
            print(key,"Even")
    else:
        print(key,"Odd")
```

```python
#Question->6. Shopping Cart Classification
products={
    "Laptop" : 15000,
    "Mouse" : 500,
    "Keyboard" : 60000,
    "Monitor" : 1200,
    "Printer" : 8000
}
for pro_type, price in products.items():
    if(price>=1000):
        if(price>=5000):
            print(pro_type,"is Premium Product")
        elif(price<1000):
            print(pro_type,"is Affordable Product")
    else:
        print(pro_type,"is Expensive Product")
```

```python
#Quetion->7. Student Eligibility
students ={
    "Rahul": {"attendance" : 85, "marks": 70},
    "Amit": {"attendance" : 60, "marks" : 80},
    "Priya": {"attendance" : 90, "marks": 35},
    "Neha": {"attendance" : 78, "marks" : 55}
}
for name, info in students.items():
    attendance=info["attendance"]
    marks=info["marks"]
    if(attendance>=75):
        if(marks>=40):
            print(name, "Eligible")
        elif(marks<40):
            print(name, "Failed in Exam")
    else:
        print(name,"not Eligible")
    
```

```python
#Question->8. Bank account classification
accounts ={
    "Rahul" : 25000,
    "Amit" : 5000,
    "Priya" : 0,
    "Neha" : -2000
}
for name, balance in accounts.items():
    if(balance==0):
        if(balance>=10000):
            print(name, "Premium Account")
        elif(balance<0):
            print(name, "Overdraft")
        else:
            print(name,"Regular Account")
    else:
        print(name,"Account bance is zero")
            
```

```python
# Question->9. Nested Dictionary - Employee Experience
employees = {
    "Rahul": {"salary" : 60000, "experience" : 8},
    "Amit": {"salary" : 40000, "experience" : 2},
    "Priya": {"salary" : 70000, "experience" : 0},
    "Neha": {"salary" : 45000, "experience" : 4}
}
for name, info in employees.items():
    salary=info["salary"]
    experience=info["experience"]
    if(salary>=50000):
        if(experience>=5):
            print(name, "is Senior")
        else:
            print(name, "is Mid Level")
    if(salary<50000):
        if(experience>=3):
            print(name, "is Experienced Junior")
        else:
            print(name, "is Fresher / Junior")
    
```

```python
# Question->10. Students grading system 
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, info in students.items():
    marks = info["marks"]
    attendance = info["attendance"]
    if(attendance>=75):
        if(marks >= 80):
            print(name , "Distinction")
        elif(marks >= 60):
            print(name , "First Division")
        elif(marks >= 40):
            print(name , "Pass")
        else:
            print(name , "Fail")
    else:
        print(name, "Not Eligible")
```
