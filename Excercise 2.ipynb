```python
# Question 2. Employee Bonus

exp=int(input("Enter Your Experience:"))
if(exp>=1):
    salary=int(input("Enter Your Salary:"))
    per=int(input("Enter Your Performance Score:"))
    if(salary>=50000):
        if(per>=90):
            bonus=salary*20/100
            salary+=bonus
            print("Your Bonus is (20%) :",bonus)
            print("Your Salary after Bonus:",salary)
        elif(per>=75):
            bonus=salary*10/100
            salary+=bonus
            print("Your Bonus is (10) :",bonus)
            print("Your Salary after Bonus:",salary)
        else:
            bonus=salary*5/100
            salary+=bonus
            print("Your Bonus is (5%) :",bonus)
            print("Your Salary after Bonus:",salary)
    else:
        if salary<50000:
            if per>=90:
                bonus=salary*15/100
                salary+=bonus
                print("Your Bonus is (15%):",bonus)
                print("Your Salary after Bonus:",salary)
            else:
                bonus=salary*5/100
                salary+=bonus
                print("Your Bonus is (5%) :",bonus)
                print("Your Salary after Bonus:",salary)
else:
    print("Experience is Zero")
          
```
