```python
# 2. Shopping Discount Calculator
def calculate_discount(bill_amount, discount=None):
    if bill_amount <= 0:
        print("Invalid Bill Amount")
        return
    if discount is not None:
        if not (0 <= discount <= 100):
            print("Invalid Custom Discount! Must be between 0% and 100%.")
            return
        discount_percentage = discount
    else:
        if bill_amount < 1000:
            discount_percentage = 5
        elif 1000 <= bill_amount < 4999:
            discount_percentage = 10
        elif 5000 <= bill_amount < 9999:
            discount_percentage = 15
        else:  
            discount_percentage = 20

    discount_amount = (discount_percentage / 100) * bill_amount
    final_payable = bill_amount - discount_amount


    print(f"Original Bill      : ₹{bill_amount:.2f}")
    print(f"Discount Percentage: {discount_percentage}%")
    print(f"Discount Amount    : ₹{discount_amount:.2f}")
    print(f"Final Payable      : ₹{final_payable:.2f}")
    print("-" * 30)


calculate_discount(600)        
calculate_discount(5500)       
calculate_discount(2000, 25)   
calculate_discount(-50)       
```

```python
# Excercise 6.
def calculate_Avg(*marks):
    if marks is not None:
        print("Marks :", marks)
        avg = sum(marks)/len(marks)
        print("Average Marks :", avg)
    else:
    
        print("Invalid Marks!")


marks = [45,56,45,77,95,37,94,44,22]
calculate_Avg(*marks)
    
```

```python
def calculate_avg(marks):
    if not marks:
        print("Error! : Input list is empty.")
        return None

    valid_marks = []

    for mark in marks :
        if not isinstance(mark, (int, float)):
            print(f"Rejectetd Invalid Type : {mark}")
            continue
        if mark < 0 or mark > 100:
            print(f"Rejected out-of-range marks : {marks}")
            continue

        valid_marks.append(mark)

    if not valid_marks:
        print("Error! : No valid marks left to compute average.")
        return None

    avg = sum(valid_marks)/len(valid_marks)
    print(f"Successfully Processed {len(valid_marks)} subjects.")
    print(f"Calculate Average : {avg : .2f}")
    

test_marks = [45, 56.45, 87 ,90.7 ,24]

calculate_avg(test_marks)
```
